# Building GPT-2 from Scratch in PyTorch

This interactive notebook demonstrates the core components of OpenAI's GPT-2 architecture step by step:
1. **Token and Positional Embeddings**
2. **Causal Self-Attention**
3. **Feed-Forward MLP Network**
4. **Transformer Decoder Block**
5. **Full GPT-2 Model & Autoregressive Text Generation**

```python
import sys
sys.path.append('../src')

import torch
import torch.nn as nn
from gpt2.config import GPTConfig
from gpt2.model import GPT, Block
from gpt2.attention import CausalSelfAttention
from gpt2.mlp import MLP
from gpt2.tokenizer import GPT2Tokenizer

print(f"PyTorch version: {torch.__version__}")
```

## 1. Configurations and Architecture Sizes

GPT-2 comes in four standard sizes: Small (124M), Medium (355M), Large (774M), and XL (1558M).

```python
config = GPTConfig.gpt2_124m()
print(config)
```

## 2. Inspecting Causal Self-Attention

The attention mechanism projects inputs into Query, Key, and Value representations and applies a causal mask so earlier tokens cannot look into the future.

```python
attn = CausalSelfAttention(config)
x = torch.randn(2, 16, config.n_embd)  # (Batch=2, SeqLen=16, EmbedDim=768)
out = attn(x)
print(f"Attention input shape:  {x.shape}")
print(f"Attention output shape: {out.shape}")
```

## 3. Initializing and Testing Full Model

```python
model = GPT(config)
print(f"Total parameters: {model.get_num_params():,}")

tokenizer = GPT2Tokenizer()
prompt = "Artificial intelligence will"
tokens = tokenizer.encode(prompt)
idx = torch.tensor(tokens).unsqueeze(0)

generated_tokens = model.generate(idx, max_new_tokens=20, temperature=0.8)
print(tokenizer.decode(generated_tokens[0].tolist()))
```
